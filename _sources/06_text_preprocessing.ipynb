# Chương 3: Khám phá & Tiền xử lý Dữ liệu Văn bản

> *Trạng thái: Khung chương khởi tạo (Giai đoạn 1). Nội dung EDA và pipeline NLP sẽ được triển khai ở Giai đoạn 3.*

## Mục tiêu chương
- Text EDA: Phân phối độ dài văn bản + kiểm định Mann-Whitney U, Word Cloud, Top-50 từ, TTR và phân tích định luật Zipf.
- Pipeline chuẩn hóa 5 bước với thống kê biến động từ vựng & độ dài sau từng bước.
- So sánh 4 chiến lược Tokenization (Word-level, Sentence-level, Character-level, Subword BPE) qua Vocab size, OOV rate và độ dài chuỗi.
- Loại bỏ Stop words (MI & Naive Bayes) và so sánh Stemming vs Lemmatization (Porter, Snowball, WordNet; Collision rate & 5-fold CV).
- Vector hóa (BoW, TF-IDF n-gram, Word2Vec) và `[Nâng cao]` Sentence Transformer (`all-MiniLM-L6-v2`) kèm K-Means Silhouette & Linear SVM F1-score.
