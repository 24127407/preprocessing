# Chương 1: Khám phá Dữ liệu Bảng (Tabular EDA)

> *Trạng thái: Khung chương khởi tạo (Giai đoạn 1). Nội dung EDA chuyên sâu sẽ được triển khai ở Giai đoạn 2.*

## Mục tiêu chương
- Kiểm định phân phối chuẩn (Shapiro-Wilk / D'Agostino-Pearson) cho toàn bộ thuộc tính số.
- Phân tích tương quan đa biến (Pearson & Spearman heatmap) và xác định cặp đa cộng tuyến mạnh ($|r| > 0.9$).
- Trực quan hóa ma trận khuyết thiếu (`missingno`), kiểm định Little's MCAR test và biện luận cơ chế MCAR/MAR/MNAR.
